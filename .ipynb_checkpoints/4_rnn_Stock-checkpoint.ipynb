```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, confusion_matrix, accuracy_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
df = pd.read_csv("4_rnn_stock.csv")
data = df[['Close']].values
```

```python
scaler = MinMaxScaler() # scale all data points to a range between 0 and 1
# The minimum value in your dataset becomes exactly 0.0.
# The maximum value in your dataset becomes exactly 1.0.
data = scaler.fit_transform(data)
```

```python
# ================= CREATE SEQUENCES =================
# This code block prepares the sequential data in a supervised learning format. It creates a sliding window of historical data, which is the exact input format required to train a recurrent neural network (RNN).
X, y = [], []
# We start at 60 because we need at least 60 prior days of data before we can make our very first prediction.
# Sliding Window Logic
for i in range(60, len(data)):
    X.append(data[i-60:i])
    y.append(data[i])

X = np.array(X)
y = np.array(y)
```

```python
# ================= TRAIN TEST SPLIT =================
# alculates the split index. It multiplies the total number of samples (len(X)) by 0.8 to allocate 80% of the data for training and the remaining 20% for testing.
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# we must never shuffle time series data
```

```python
model = Sequential([
    SimpleRNN(64, return_sequences=True, input_shape=(60,1)),
    Dropout(0.2),
    SimpleRNN(64),
    Dropout(0.2),
    Dense(25, activation='relu'),
    Dropout(0.2),

    Dense(1)
])

model.compile(optimizer='adam', loss='huber') # it acts like MSE for small errors, but like Mean Absolute Error (MAE) for large errors. This prevents extreme outliers
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_test, y_test),
    callbacks=[early_stop],
    verbose=1 # Displays an active training progress bar
) 
```

```python
pred = model.predict(X_test)
# reverts the scaling (normalization) on both the predictions and actual values so they are readable in their original units
pred = scaler.inverse_transform(pred)
real = scaler.inverse_transform(y_test)
```

```python
# ================= METRICS =================
rmse = np.sqrt(mean_squared_error(real, pred))
print("RMSE :", rmse)
```

```python
# ================= ACTUAL VS PREDICTED =================

plt.figure(figsize=(12,5))

plt.plot(real, label='Actual')
plt.plot(pred, label='Predicted')

plt.title("Actual vs Predicted")
plt.legend()
plt.show()

# ================= LOSS GRAPH =================

plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title("Loss Graph")
plt.legend()
plt.show()
```
